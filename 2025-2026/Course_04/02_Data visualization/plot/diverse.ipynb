## Setting Axis Limits Using Python

```python
# Setting Axis Limits Using Python 
import matplotlib.pyplot as plt
from numpy import sin
import numpy as np
x = np.linspace(0, 10, 100)
y=sin(x)
plt.plot(x, y) 
plt.xlim(0, 7) 
plt.ylim(-1, 1)
plt.show()
```

### Sunburst Chart in Python

```python
import plotly.express as px

# Sample data
data = {
    "id": ["A", "B", "C", "D", "E", "F"],
    "parent": ["", "A", "A", "B", "B", "C"],
    "value": [16, 25, 7, 8, 12, 6]
}
# Create a sunburst chart 
fig = px.sunburst(data, names='id', parents='parent', values='value' )

# Set the chart title 
fig.update_layout(title_text="Sunburst Chart")

# Show the chart
fig.show() 
```

```python
# Wireframe plot 
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Create data
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
x, y = np.meshgrid(x, y)
z = np.sin(np.sqrt(x**2 + y**2))

# Create a 3D plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.set_title('3D Wireframe Plot')
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')

# Plot wireframe
ax.plot_wireframe(x, y, z, color='green')
plt.show()
```

## Candlestick Chart plot using in Python

```python
# Candlestick Chart plot using in Python
import yfinance as yf
import plotly.graph_objects as go
# Download historical data for Apple Inc. (AAPL)
data = yf.download('AAPL', start='2025-01-01', end='2025-12-31')
# Create a candlestick chart
fig = go.Figure(data=[go.Candlestick(x=data.index,
                open=data['Open'],
                high=data['High'],
                low=data['Low'],
                close=data['Close'])])

# Set the chart title
fig.update_layout(title_text='AAPL Candlestick Chart')

# Show the chart
fig.show()
```

## Python code to plot Sigmoid Activation Functions

```python
# Python code to plot Sigmoid Activation Functions
import numpy as np 
import matplotlib.pyplot as plt
# define sigmoid(x) return 1 / (1 + np.exx(-x)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
# Input range
x = np.linspace(-10, 10, 500)
plt.figure(figsize=(10, 6))
plt.plot(x, sigmoid(x), label="Sigmoid")
plt.title("Activation Function", fontsize=16)
plt.xlabel("Input", fontsize=14)
plt.ylabel("Output", fontsize=14)
plt.axhline(0, color="black", linewidth=0.5, linestyle="-")
plt.axvline(0, color="black", linewidth=0.5, linestyle="-")
plt.legend(fontsize=12)
plt.grid(alpha=0.3)
plt.show()
```

## 3D Plot using Python

```python
# 3D Plot using Python

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# data
x=[1,2,3,4,5]
y=[2,3,4,5,6]
z=[10,5,4,8,9]

# plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(x, y, z, color='red')
ax.set_title('3D Scatter Plot')
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')
plt.show()
```
