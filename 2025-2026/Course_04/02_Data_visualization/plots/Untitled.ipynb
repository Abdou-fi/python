```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(-5, 5, 0.01)
# y = np.sqrt(x)
z = np.cos(x)
h = np.sin(x)
# plt.plot(x, y)
plt.plot(x, z)
plt.plot(x, h)
plt.show()
```

```python
import numpy as np 
import matplotlib.pyplot as plt 
x = np.arange(-5, 5, 0.001)
y = np.sin(x)
plt.plot(x, y)
plt.xlim(0, 5)
plt.ylim(-1, 1)
plt.show()
```

```python
import matplotlib.pyplot as pl  
x = [ '1',  2,  3,  4,  5]
z = [10, 12, 15, 17, 20]

pl.plot(x, z,'-go')  # Plot x and z with green circle markers on each point
pl.title('evolution of salary over months')
pl.xlabel('months')
pl.ylabel('salary')
pl.grid(axis='y')
pl.show()
```
