
### Customizing plots

You can customize the appearance of your plots using various options. For example, you can change the line color, line width, marker style, and more:

```python
import matplotlib.pyplot as plt

# Create some data
x = ['2021', 2022, 2023, 2024, 2025]
ESS = [71, 54, 42, 48, 41]
MCA = [54, 51, 47, 65, 58]

# Create a plot with custom options
plt.plot(x, ESS, color='black', linewidth=1, marker='o', markersize=10, markerfacecolor='white', linestyle='--')  
plt.plot(x, MCA, color='green', linewidth=1, marker='*', markersize=10, markerfacecolor='red')
plt.title('ESS and MCA Points Over Time')
plt.xlabel('Year')
plt.ylabel('Points')
plt.legend(['ESS', 'MCA'])

# Show the plot
plt.show()
```

```python
import yfinance as yf
import matplotlib.pyplot as plt
# Download historical data for Apple Inc. (AAPL)
data = yf.download('AAPL', start='2023-01-01', end='2025-12-31')
# Plot the closing prices
plt.figure(figsize=(10, 5))
plt.plot(data.index, data['Close'], label='AAPL Closing Prices')
plt.title('Apple Inc. (AAPL) Closing Prices in 2023-2025')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend()
plt.grid()  
plt.show()
```

## Horizontal Bar Graph:

```python
import matplotlib.pyplot as pyplot # Set up the data
labels = ('Python', 'Scala', 'C#', 'Java', 'PHP') 
index = (1, 2, 3, 4, 5) # provides Locations on x axis 
sizes = [45, 10, 15, 30, 22]

# Set up the bar chart
pyplot.barh(index, sizes, tick_label=labels, color=('red','green', 'blue', 'yellow', 'orange'))

# Configure the Layout
pyplot.ylabel("Usage")
pyplot.xlabel("Programming Languages")

# Display the chart 
pyplot.show()
```

## Stacked Bar Graph:

```python
import matplotlib.pyplot as plt

# Set up the data
labels=['Python', 'Scala', 'C#', 'Java', 'PHP']
index = [1, 2, 3, 4, 5] # provides Locations on x axis
web_usage = [20, 2, 5, 10, 14]
data_science_usage = [25, 8, 5, 15, 2]
games_usage = [10, 1, 5, 5, 4]

# Set up the bar chart
plt.bar(index, web_usage, tick_label=labels, label='web') 
plt.bar(index, data_science_usage, tick_label=labels, label='data science', bottom=web_usage)
web_and_games_usage = [web_usage[i] + data_science_usage[i] for i in range(0, len(web_usage))]
plt.bar(index, games_usage, tick_label=labels, label='games', bottom=web_and_games_usage)

# Configure the Layout
plt.ylabel('Usage')
plt.title('Programming Language Usage by Domain')
plt.legend()

# Display the chart
plt.show()
```

### 4. Bokeh

```python
# Bar graph plot using different Python libraries
# 4. Bokeh
from bokeh.plotting import figure, show
from bokeh.io import output_notebook

# sample data
categories = ['A','B', 'C', 'D']
values = [10, 25, 30, 15]

# Create a bar chart using Bokeh
p=figure(x_range=categories, title="Bar Chart Example", x_axis_label='Categories', y_axis_label='Values')
p.vbar(x=categories, top=values, width=0.5)

# Display the chart
output_notebook()
show(p)
```

```python
# Bar graph plot using different Python libraries
# 5. Altair
import altair as alt
import pandas as pd

# Sample data
data = pd.DataFrame({
    "categories": ['Category 1', 'Category 2', 'Category 3', 'Category 4'],
    "values": [10, 25, 15, 30]
})

# Create a bar chart using Altair
chart = alt.Chart(data).mark_bar().encode(
    x='categories',
    y='values'
).properties(
    title='Bar Chart Example'
)   
chart.show()
```

## Grouped Bar Graph

```python
import matplotlib.pyplot as pyplot
BAR_WIDTH = 0.35
# set up grouped bar charts 
teama_results = (66, 75, 56, 62, 58) 
teamb_results = (55, 68, 80, 73, 55)

# Set up the index for each bar 
index_teama = (1, 2, 3, 4, 5) 
index_teamb = [i + BAR_WIDTH for i in index_teama]

# Determine the mid point for the ticks 
ticks = [i + BAR_WIDTH / 2 for i in index_teama] 
tick_labels = ('Lab 1', 'Lab 2', 'Lab 3', 'Lab 4', 'Lab 5')

# Plot the bar charts
pyplot.bar(index_teama, teama_results, BAR_WIDTH, color="b", label='Team A')
pyplot.bar(index_teamb, teamb_results, BAR_WIDTH, color="g", label='Team B')

# Configure the layout
pyplot.xlabel('Labs')
pyplot.ylabel('Scores')
pyplot.title('Team Results by Lab')
pyplot.xticks(ticks, tick_labels)
pyplot.legend()

# Display the chart
pyplot.show()
```

## Animated scatter plot using Python

```python
import matplotlib.pyplot as plt 
import matplotlib.animation as animation
import numpy as np
num_points = 100
x, y = np.random.rand(2, num_points) * 10 
colors = np.random.rand(num_points)
sizes = np.random.rand(num_points) * 1000
fig, ax = plt.subplots()
scat = ax.scatter(x, y, c=colors, s=sizes) 
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
def animate(i): 
    scat.set_offsets(np.c_[x + np.random.randn(num_points) * 0.1, y + np.random.randn(num_points) * 0.1]) 
    return scat,
ani = animation.FuncAnimation(fig, animate, frames=100, interval=50) 
plt.show()
```

## 11. Radar Chart:

```python
import matplotlib.pyplot as plt
import numpy as np

labels = ['A', 'B', 'C', 'D', 'E']
data = np.array([4, 5, 3, 4, 2])

angles = np.linspace(0, 2*np.pi, len(labels), endpoint=False)
data = np.concatenate((data, [data[0]]))  # Close the loop
angles = np.concatenate((angles, [angles[0]]))

plt.polar(angles, data, marker='o')
plt.title('Radar Chart')
# plt.xticks(angles[:-1], labels)
plt.show()
```

## Timeline Graphics

Timeline graphics in Python can be created using several libraries, with Plotly and Matplotlib being the most common choices. Plotly offers interactive, web-ready visualizations, while Matplotlib is a powerful tool for static plots, including specialized timeline examples in its gallery. 

<img src="images/TimeLine.png" alt="interactive" width="200"/>

### 1. Using plotly (interactive Timeline/Gantt Chart):

```python
import plotly.express as px
import pandas as pd
import datetime as dt

df = pd.DataFrame([
    dict(Task="Project Planning", Start='2025-01-10', Finish='2025-02-15'),
    dict(Task="Development", Start='2025-02-15', Finish='2025-03-01'),
    dict(Task="Testing", Start='2025-03-01', Finish='2025-03-05'),
    dict(Task="Launch", Start='2025-03-05', Finish='2025-03-10')
])

df['Start'] = pd.to_datetime(df['Start'])
df['Finish'] = pd.to_datetime(df['Finish'])

fig = px.timeline(df, x_start="Start", x_end="Finish", y="Task", color="Task")
fig.update_yaxes(autorange="reversed") # optional: tasks are listed from top to bottom
fig.show()


```

### 2. Using Matplotlib (static Timeline):


```python
# 2. Using Matplotlib (static Timeline):
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import pandas as pd
import datetime as dt
# Sample data
data = {
    "Task": ["Project Planning", "Development", "Testing", "Launch"],
    "Start": [dt.datetime(2025, 1, 10), dt.datetime(2025, 2, 15), dt.datetime(2025, 3, 1), dt.datetime(2025, 3, 5)],
    "Finish": [dt.datetime(2025, 2, 15), dt.datetime(2025, 3, 1), dt.datetime(2025, 3, 5), dt.datetime(2025, 3, 10)]
}
df = pd.DataFrame(data)
# Plot
fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(df["Task"], df["Finish"] - df["Start"], left=df["Start"])
ax.xaxis_date()
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
ax.set_xlabel('Date')
ax.set_ylabel('Task')
ax.set_title('Timeline')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
