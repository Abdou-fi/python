## Top Python Packages for Data Visualization

## Subplots

Matplotlib allows you to create multiple plots within the same figure. In order to add multiple plots,
you need to create a “subplot” for each plot in the figure you’d like to use.

This is done with the `add_subplot()` function, which accepts a series of numeric arguments.

The first number specifies how many rows you want to add to the figure, the second number specifies
how many columns you want to add, and the third number specifies the number of the plot that
you want to add.

This means that if you in passed in `111` into the `add_subplots()` function, one new subplot would
be added to the figure. Meanwhile, if you used the numbers `221`, the resulting plot would have four
axes with two columns and two rows - and the subplot you’re forming is in the 1st position.

Here’s how we would create two subplots in the same figure, notice that we have created two axes
objects:

````python
import matplotlib.pyplot as plt 
fig = plt.figure()
names = ['2021', '2022', '2023']
values_1 = [19, 50, 29]
values_2 = [48, 19, 41]
    
ax1 = fig.add_subplot(121)       # 1 row, 2 columns, subplot 1
ax2 = fig.add_subplot(122)      # 1 row, 2 columns, subplot 2

ax1.bar(names, values_1)
ax2.bar(names, values_2)
plt.show()
````

We’ve created two sublots in a figure with 1 row and 2 columns. They’re sitting side by side. If we
had created a figure with 2 rows and 1 column:

````python
fig = plt.figure()

names = ['A', 'B', 'C']
values = [19, 50, 29]
values_2 = [48, 19, 41]
values_3 = [25, 40, 30]

ax1 = fig.add_subplot(311)       # 3 rows, 1 column, subplot 1
ax2 = fig.add_subplot(312)      # 3 rows, 1 column, subplot 2
ax3 = fig.add_subplot(313)      # 3 rows, 1 column, subplot 3
ax1.bar(names, values)
ax2.bar(names, values_2)
ax3.bar(names, values_3)
plt.savefig("plot_subplots.svg")
plt.show()
````

## Stacked Bar Graph:

````python
import matplotlib.pyplot as plt

# Set up the data
labels=['Python', 'Scala', 'C#', 'Java', 'PHP']
index = [1, 2, 3, 4, 5] # provides Locations on x axis
web_usage = [20, 2, 5, 10, 14]
data_science_usage = [25, 8, 5, 15, 2]
games_usage = [10, 1, 5, 5, 4]

# Set up the bar chart
plt.bar(index, web_usage, tick_label=labels, label='web') 
plt.bar(index, data_science_usage, tick_label=labels, label='data-science', bottom=web_usage)
web_and_data_science_usage = [web_usage[i] + data_science_usage[i] for i in range(0, len(web_usage))]
plt.bar(index, games_usage, tick_label=labels, label='games', bottom=web_and_data_science_usage)

# Configure the Layout
plt.ylabel('Usage %')
plt.title('Programming Language Usage by Domain')
# set legend transparency to 50%
plt.legend(loc='best', framealpha=0.3)


# Display the chart
plt.show()
````

## Grouped Bar Graph

````python
import matplotlib.pyplot as pyplot
BAR_WIDTH = 0.35
# set up grouped bar charts 
teama_results = (66, 75, 56, 62, 58) 
teamb_results = (55, 68, 80, 73, 55)

# Set up the index for each bar 
index_teama = (1, 2, 3, 4, 5) 
index_teamb = [i + BAR_WIDTH for i in index_teama]

# Determine the mid point for the ticks 
ticks = [i + BAR_WIDTH / 2 for i in index_teama] 
tick_labels = ('year 1', 'year 2', 'year 3', 'year 4', 'year 5')

# Plot the bar charts
pyplot.bar(index_teama, teama_results, BAR_WIDTH, color="#950F83", label='Team A')
pyplot.bar(index_teamb, teamb_results, BAR_WIDTH, color="g", label='Team B')

# Configure the layout
pyplot.xlabel('Labs')
pyplot.ylabel('Scores')
pyplot.title('Team Results by Lab')
pyplot.xticks(ticks, tick_labels)
pyplot.legend()

# Display the chart
pyplot.show()
````

## Timeline Graphics

Timeline graphics in Python can be created using several libraries, with Plotly and Matplotlib being the most common choices. Plotly offers interactive, web-ready visualizations, while Matplotlib is a powerful tool for static plots, including specialized timeline examples in its gallery. 

<img src="images/TimeLine.png" alt="interactive" width="200"/>

### 1. Using plotly (interactive Timeline/Gantt Chart): 


````python
import plotly.express as px
import pandas as pd
import datetime as dt

df = pd.DataFrame([
    dict(Task="Project Planning", Start='2025-01-10', Finish='2025-02-15'),
    dict(Task="Development", Start='2025-02-15', Finish='2025-03-01'),
    dict(Task="Testing", Start='2025-03-01', Finish='2025-03-05'),
    dict(Task="Launch", Start='2025-03-05', Finish='2025-03-10')
])

df['Start'] = pd.to_datetime(df['Start'])
df['Finish'] = pd.to_datetime(df['Finish'])

fig = px.timeline(df, x_start="Start", x_end="Finish", y="Task", color="Task")
fig.update_yaxes(autorange="reversed") # optional: tasks are listed from top to bottom
fig.show()


````

### 2. Using Matplotlib (static Timeline):


````python
# 2. Using Matplotlib (static Timeline):
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import pandas as pd
import datetime as dt
# Sample data
data = {
    "Task": ["Project Planning", "Development", "Testing", "Launch"],
    "Start": [dt.datetime(2025, 1, 10), dt.datetime(2025, 2, 15), dt.datetime(2025, 3, 1), dt.datetime(2025, 3, 5)],
    "Finish": [dt.datetime(2025, 2, 15), dt.datetime(2025, 3, 1), dt.datetime(2025, 3, 5), dt.datetime(2025, 3, 10)]
}
df = pd.DataFrame(data)
# Plot
fig, ax = plt.subplots(figsize=(10, 8))
ax.barh(df["Task"], df["Finish"] - df["Start"], left=df["Start"])
ax.xaxis_date()
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
ax.set_xlabel('Date')
ax.set_ylabel('Task')
ax.set_title('Timeline')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

# Funnel charts in Python 
Funnel charts can be created using libraries such as Matplotlib and Plotly. These charts are useful for visualizing the reduction of a process or a sequence of steps, where each step represents a stage in the process and the size of the step represents the number of items that move through that stage.

````python
import plotly.graph_objects as go
stages = ["Sent", "Viewed", "clicked", "add to cart","purchased"] 
values = [1000, 700, 400, 200, 100]
fig = go.Figure(go.Funnel(
    y=stages, 
    x=values
))
fig.update_layout(title="website Conversion Funnel")
fig.show()
````

# Donut charts in Python

Donut charts can be created using libraries such as Matplotlib and Plotly. These charts are similar to pie charts but have a hole in the center, making them more visually appealing and easier to read.
Hollowing out the center strips away confusing angles. It forces the eye to compare arc lengths, instantly reducing visual clutter.

````python
# Donut charts in Python
import matplotlib.pyplot as plt
labels = ['A', 'B', 'C', 'D']
sizes = [15, 30, 45, 10]
colors = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue']
explode = (0, 0, 0, 0)  # explode 1st slice (i.e., 'A')
fig1, ax1 = plt.subplots()
ax1.pie(sizes, explode=explode, labels=labels, colors=colors,
        autopct='%1.1f%%', startangle=90, wedgeprops={'width': 0.3}, counterclock=False)  # Set width to create a donut shape
ax1.set_title('Donut Chart Example')
plt.show()
````

# Stucked area chart in Python

Stuck area charts can be created using libraries such as Matplotlib and Plotly. These charts are useful for visualizing the cumulative effect of a series of values over time, where the area between the line and the x-axis represents the cumulative sum of the values.

````python
# stucked area chart in Python

import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales_a = [100, 120, 150, 130, 170, 160]
sales_b = [80, 90, 100, 110, 120, 130]
sales_c = [60, 70, 80, 90, 100, 110]
plt.stackplot(months, sales_a, sales_b, sales_c, colors=['blue', 'orange', 'green'])

plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Stacked Area Chart Example')
plt.legend(["Sales A", "Sales B", "Sales C"], loc='upper left')

plt.show()
````

# Sunburst in Python

Sunburst charts can be created using libraries such as Matplotlib and Plotly. These charts are useful for visualizing hierarchical data, where each level of the hierarchy is represented as a ring, with the size of the ring representing the value of that level.

````python
# sunburst chart in Python
import plotly.express as px
data = dict(
    character=["A", "B", "C", "D", "E", "F"],
    parent=["", "A", "A", "B", "B", "C"],
    value=[10, 20, 30, 40, 50, 60]
)
fig = px.sunburst(data, names="character", parents="parent", values="value")
fig.show()
````

````python
# sunburst chart in Python
import plotly.express as px
import pandas as pd
data = pd.DataFrame({
    "category": ["Electronics", "Electronics", "Electronics", "Clothing", "Clothing", "Clothing"],
    "subcategory": ["Mobiles", "Laptops", "Accessories", "Men", "Women", "Babies"],
    "value": [40, 30, 20, 20, 10, 15]
})

fig = px.sunburst(data, 
                  path=["category", "subcategory"], 
                  values="value", 
                  title="Sales distribution by category")
fig.show()
````

# Treemap chart in Python

Treemap charts can be created using libraries such as Matplotlib and Plotly. These charts are useful for visualizing hierarchical data, where each level of the hierarchy is represented as a rectangle, with the size of the rectangle representing the value of that level.

````python
# Treemap chart in Python
import matplotlib.pyplot as plt
import squarify
libels =['A', 'B', 'C', 'D', 'E']
sizes = [40, 25, 15, 10, 10]

plt.figure(figsize=(8, 6))
squarify.plot(
            sizes=sizes, 
            label=libels, 
            color=plt.cm.Set3.colors,
            alpha=0.9
)
            
plt.title('Treemap Chart Example')
plt.axis('off')
plt.tight_layout()
plt.show()
````

# Timeline chart in Python

Timeline charts can be created using libraries such as Matplotlib and Plotly. These charts are useful for visualizing the progression of events over time, where each event is represented as a marker on the chart, with the x-axis representing time and the y-axis representing the events.

````python
# Timeline chart in Python

import matplotlib.pyplot as plt
import datetime as dt
tasks = ["Project Planning", "Development", "Testing", "Launch"]
dates=[dt.datetime(2025, 1, 10), dt.datetime(2025, 2, 15), dt.datetime(2025, 3, 1), dt.datetime(2025, 3, 5)]

y=[1,1,1,1]

plt.scatter(dates, y)
for i, task in enumerate(tasks):
    plt.text(dates[i], y[i]+0.02, task, rotation = 45, ha='right')
plt.yticks([])
plt.xlabel('Timeline')
plt.title('Timeline Chart Example')

plt.show()

````

# Stream graph in Python

Stream graphs can be created using libraries such as Matplotlib and Plotly. These charts are useful for visualizing the flow of data over time, where each data point is represented as a line segment, with the x-axis representing time and the y-axis representing the value of the data point.


````python
# Stream graph in Python

import matplotlib.pyplot as plt
import numpy as np

x = np.arange(10)
y1=np.random.rand(10)
y2=np.random.rand(10)
y3=np.random.rand(10)

y=np.vstack([y1, y2, y3])
plt.stackplot(x,y, baseline='wiggle', colors=['blue', 'orange', 'green'] )

plt.xlabel('Time')
plt.ylabel('Value')
plt.title('Stream Graph Example')
plt.legend(['Stream A', 'Stream B', 'Stream C'])

plt.show()
````

# Density plot in Python

Density plots can be created using libraries such as Matplotlib and Plotly. These charts are useful for visualizing the distribution of data, where the area under the curve represents the proportion of data points that fall within a given range.



````python
# Density plot in Python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

data = np.random.normal(0, 1, 1000)
sns.kdeplot(data, fill=False, color="blue", bw_adjust=0.9)

plt.title('Density Plot Example')
plt.xlabel('Value')
plt.ylabel('Density')
plt.show()
````

# mixed plots

Mixed plots can be created using libraries such as Matplotlib and Plotly. These charts combine different types of plots, such as line plots, bar plots, and scatter plots, to provide a more comprehensive view of the data.

````python
# bar and line plot in Python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
# bar graph
categories = ['A','B', 'C', 'D']
values = [25, 40, 30, 20]
plt.bar(categories, values)

# line graph
x = np.arange(len(categories))
y = np.array(values)
plt.plot(x, y, color='red', marker='o', label='Line Graph')
plt.title('Bar and Line Plot Example')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.legend()
plt.show()


````

# Strip plot in Python

Strip plots can be created using libraries such as Matplotlib and Plotly. These charts are useful for visualizing the distribution of data, where each data point is represented as a line segment, with the x-axis representing the data points and the y-axis representing the value of the data point.

````python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="white", font='serif')
plt.figure(figsize=(10, 6), facecolor='lightgray')

df = sns.load_dataset("trips")
ax = sns.lineplot(x="day", y="total_bill", data=df,
                                     jitter=0.25, size = 8, alpha=0.6,
                                     palette= [ "#ES9898", "#858380", "#6D6875", "#D8C1AD"])

ax.set_facecolor('#FAF9F6')
sns.despine(left=True, bottom=True) 
plt.title("Daily spending flow", fontsize=18, pad= 20, color='#4A4A4A' )
plt.xlabel("")
plt.ylabel("Amount ($)", fontsize=12, color='#6D6875')
plt.show()
````

# The waffle chart
Waffle charts can be created using libraries such as Matplotlib and Plotly.
Shows parts-of-a-whole but uses a grid. Aesthetic and easy to read percentages.

````python
import matplotlib.pyplot as plt
from pywaffle import Waffle

# Data represented as a dictionary
data = {'Cats': 50, 'Dogs': 28, 'Rabbits': 19}

# Create the waffle chart
fig = plt.figure(
    FigureClass=Waffle,
    rows=5, 
    values=data, 
    legend={'loc': 'upper left', 'bbox_to_anchor': (1.05, 1)},
    title={'label': 'Pet Popularity', 'loc': 'center'}
)

plt.show()
````

## Waffle customization 

A waffle chart shows progress towards a target or a completion percentage. There is a grid of small cells, of which coloured cells represent the data.

A chart can consist of one category or several categories. Multiple waffle charts can be put together to show a comparison between different charts.

````python
import matplotlib.pyplot as plt
from pywaffle import Waffle
import pandas as pd

# 1. Prepare data
# The total sum of values should equal 100 for a 10x10 grid (1 tile = 1%)
data = {'Category A': 45, 'Category B': 30, 'Category C': 25}

# 2. Create the Figure
fig = plt.figure(
    FigureClass=Waffle,
    rows=10,                      # 10 rows
    columns=10,                   # 10 columns
    values=data,                  # Data values
    cmap_name="tab10",            # Color scheme
    legend={'loc': 'upper left', 'bbox_to_anchor': (1, 1)}, # Legend location
    title={'label': '10x10 Waffle Chart', 'loc': 'center'},
    figsize=(7, 5)                # Figure size
)

plt.show()
````

````python
import matplotlib.pyplot as plt
from pywaffle import Waffle

# Data: Category Value (e.g., 75%) and Remaining Value (25%)
data = {'Category A': 75, 'Remaining': 25}

# Create Figure
fig = plt.figure(
    FigureClass=Waffle,
    rows=10,                      # 10 rows for 10x10
    columns=10,                   # 10 columns for 10x10
    values=data,
    colors=["#232066", "#eeeeee"], # Color for Category, Color for Remaining
    title={'label': '10x10 Waffle Chart', 'loc': 'left'},
    labels=[f"{k} ({v}%)" for k, v in data.items()],
    legend={'loc': 'lower left', 'bbox_to_anchor': (0, -0.2), 'ncol': len(data)}
)

plt.show()
````
