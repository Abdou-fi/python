```python
class Car:
    def __init__(self, model, year, color, for_sale, mileage=7000):
        self.model = model
        self.year = year
        self.color = color
        self.for_sale = for_sale
        self.mileage = mileage

    def get_descriptive_name(self):
        long_name = f"{self.year} {self.color} {self.model} {'for sale' if self.for_sale else 'not for sale'}"
        return long_name.title()

    def read_odometer(self):    
        print(f"This car has {self.mileage} miles on it.")


car1=Car('Toyota Camry', 2020, 'red', True)
print(car1)
print(car1.model)
print(car1.year)
print(car1.color)
print(car1.for_sale)
print(car1.mileage)


print(car1.get_descriptive_name())
print(car1.read_odometer())
car1.mileage = 15000
print(car1.read_odometer())

car2=Car('Chevrolet', 2023, 'yellow', False, 25000)
print(car2.get_descriptive_name())
print(car2.read_odometer())
```
