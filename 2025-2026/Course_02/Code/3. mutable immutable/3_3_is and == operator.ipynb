## <span class="title2"> Python <span class="orangeText">*"is"*</span> vs. <span class="orangeText">*"=="*</span></span>
What's the real deal?

### <span class="title2"><span class="orangeText">*"is"*</span> operator</span>
It checks if two variables point to the same object in memory.

Compares identity.

Often used for 'None' (e.g., 'x is None').

### <span class="title2"><span class="orangeText">*"=="*</span> operator</span>
It checks if two variables have the same value.

Compares values

Most commonly used for comparing numbers, strings, lists, etc.

Relies on the '__eq__' method of objects.

### <span class="title3">Example</span>

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)    # True
print(a is c)    # False
print(a == c)    # True
```
