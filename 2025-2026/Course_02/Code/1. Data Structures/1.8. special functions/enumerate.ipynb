## <span class="title2">enumerate</span>
The enumerate function is a built-in Python function that adds a counter to an iterable (like a list or a string) and returns it as an enumerate object. This
function is particularly useful for tracking the index of elements while iterating through a collection.

Syntax:
```python
    enumerate(iterable, start=0)
```
Parameters:
- iterable: The iterable (like a list, tuple, string, etc.) to be enumerated.
- start (optional): The starting index of the counter. Default is 0.

The enumerate function returns an enumerate object, which is an iterable that produces pairs of index and value for each element in the iterable. You can convert this object to a list of tuples using the list() function.


### <span class="title2">Example:</span>
Old traditional way 
```python
    names = ["Karim", "Amine", "Amel", "Sara", "Mohamed", "Ali"]
    for i in range (len(names)):
        print(i, names[i])
```
Pythonic way
```python
    names = ["Karim", "Amine", "Amel", "Sara", "Mohamed", "Ali"]
    for i, name in enumerate(names):
        print(i, name)       
```
Output:
```
0 Karim
1 Amine
2 Amel
3 Sara
4 Mohamed
5 Ali
```


### <span class="title2">Basic usage:</span>
```python
    my_list = ['a', 'b', 'c']
    for index, value in enumerate(my_list): 
        print(index, value)
```
Output:
```
0 a
1 b
2 c
```
### <span class="title2">Specifying a Start Index:</span>
```python
    my_list = ['a', 'b', 'c']
    for index, value in enumerate(my_list, start=1): 
        print(index, value)
```

Output:
```
1 a
2 b
3 c
```
### <span class="title2">Converting to a List of Tuples:</span>
```python
    my_list = ['a', 'b', 'c']
    enumerated list = list(enumerate(my_list))
```
Output:
```
[(0, 'a'), (1, 'b'), (2, 'c')]
```



### <span class="title2">Using with List Comprehension:</span>
```python
    enumerated _dict = {index: value for index, value in enumerate(my_list)}
```
Output:
```
{0: 'a', 1: 'b', 2: 'c'}
```

### <span class="title2">Iterating Over a String:</span>
```python
    for index, char in enumerate("hello"):
        print(index, char)
```
Output:
```
0 h
1 e
2 l
3 l
4 o
```
