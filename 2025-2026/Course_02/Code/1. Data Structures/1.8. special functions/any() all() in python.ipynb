## <span class="title2"> What do <span class="orangeText">*any( )*</span> and <span class="orangeText">*all( )*</span> do in Python?</span>
Check if ***any*** or ***all*** elements in an iterable are True

<link rel="stylesheet" type="text/css" href="style.css">

### <span class="title3">Syntax how to use:</span>
```python
    any(iterable)     # True if at least one item is True
    all(iterable)     # True if all items are True
```

### <span class="title3">When to use it?</span>
Use to quickly test multiple conditions or values in a list.

### <span class="title3">Why use it?</span>
It's simpler and faster than writing multiple if statements.
### <span class="title3">Example 1</span>

````python
nums = [0, 1, 2] 
print(any(nums))    # True 
print(all(nums))    # False
````

### <span class="title3">Example 2</span>

````python
network : dict[str : bool] = {
                "John" : False, 
                "Emily" : False, 
                "Jones" : True,
                "Nate" : False
}

# [False, False, True, False] 
if any(list(network.values())):
    print("There is at least one Person Connected")
else:
    print("No One is Connected")
````

````python
# inventory = ["chair", "desk", "lamp"]
# inventory = ["chair", None, None]
inventory = [None, None, None]

if all(inventory):
    print('Inventory full!') 
elif any(inventory):
    print('Inventory has items!')
else:
    print('Inventory is empty!')
````
