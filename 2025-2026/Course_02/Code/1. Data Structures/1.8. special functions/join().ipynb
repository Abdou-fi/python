## <span class="title2"> Using <span class="orangeText">*join( )*</span> to *combine* strings in Python</span>
What's the real deal?

The <span class="orangeText">join( )</span> method joins elements of ali st into ***a single*** string

### <span class="title3">Example</span>

````python
words = ['I', 'love', 'Python']
result = '--'.join(words)            # result = 'I love Python'
print(result)
````
