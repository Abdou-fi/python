## <span class="title2"> Using <span class="orangeText">*zip( )*</span> to *combine* lists in Python</span>
What's the real deal?

The <span class="orangeText">zip( )</span> function is a built-in function in Python that allows you to combine two or more lists into a single list of tuples.

### <span class="title3">Example</span>

```python
    countries = ['France',
                  'Germany',
                  'Greece']
    capitals = ['Paris',
                'Berlin',
                'Athens']
    for country, capital in zip(countries, capitals):
        print(f"the Capital city of {country} is {capital}")
```
Output:
```
the Capital city of France is Paris
the Capital city of Germany is Berlin
the Capital city of Greece is Athens
```
