## <span class="title2">4. Dictionaries</span>

Python dictionaries are an implementation of key-value objects, the keys being indexed.

Keys are often text, values can be of different types and structures.

To create a dictionary, you can proceed by using braces ({}). As encountered in sets, if we evaluate the following code, we get a dictionary :


```python
  empty_dict = {}
  print(type(empty_dict))
```
Output:```<class 'dict'>```

To create a dictionary with entries, the braces can be used. Each entry is separated by commas, and the key is distinguished from the associated value by two points (:):


```python
  my_dict = { "last_name": "Mohammedi",
  "first_name": "Yahia",
  "birth_year": 1992,
  "teams": ["MCA", "JSK"]}
  print(my_dict)
```
Output:```{ 'last_name': 'Mohammedi', 'first_name': 'Yahia', 'birth_year': 1992, 'teams': ['MCA', 'JSK'] }```


### <span class="title3">4.1. Accessing the values (Extraction of the Elements)</span>

To access the values associated with a key, we use the square brackets ([]):

```python
  print(my_dict["first_name"])
```
Output:```Mohammedi```

If the extraction is done by a key not present in the dictionary, an error will be returned:

```python
  print(my_dict["age"])
```
Output:
```
  ## Error in py_call_impl(callable, dots$args, dots$keywords): KeyError: 'age'
  ## 
  ## Detailed traceback: 
  ##   File "<string>", line 1, in <module> 
```

You can test the presence of a key with the operator in:

```python
  print(last_name in my_dict)
```
Output:```True```

```python
  print(age in my_dict)
```
Output:```  False```

The extraction of values can also be done using the get() method, which returns a None value if the key is not present:

```python
  print(my_dict.get("age"))
```
Output:```None```

### <span class="title3">4.2. Keys and values</span>

Using the key() method, the keys of the dictionary can be accessed:

```python
  the_keys = my_dict.keys()
  print(the_keys)
```
Output:```dict_keys(['last_name', 'first_name', 'birth_year', 'teams'])```

```python
  print(type(the_keys))
```
Output:```<class 'dict_keys'>```

It is then possible to transform this key enumeration into a list:

```python
  the_keys_list = list(the_keys())
  print(the_keys_list)
```
Output:```['last_name', 'first_name', 'birth_year', 'teams']```

The values() method provides the dictionary values:

```python
  the_values = my_dict.values()
  print(the_values)
```
Output:```dict_values(['Mohammedi', 'Yahia', 1992, ['MCA', 'JSK']])```


```python
  print(type(the_values))
```
Output:```<class 'dict_values'>```

It is then possible to transform this value enumeration into a list:

```python
  the_values_list = list(the_values())
  print(the_values_list)
```
Output:```['Mohammedi', 'Yahia', 1992, ['MCA', 'JSK']]```

The items() method provides keys and values in the form of tuples:

```python
  the_items = my_dict.items()
  print(the_items)
```
Output:```dict_items([('last_name', 'Mohammedi'), ('first_name', 'Yahia'), ('birth_year', 1992), ('teams', ['MCA', 'JSK'])])```

```python
  print(type(the_items))
```
Output:```<class 'dict_items'>```
