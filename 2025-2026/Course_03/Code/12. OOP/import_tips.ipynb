# 9 Genius Python Import Tips to Level Up Your Code
## Trick 1: import with alias

```python
import numpy as np
```

## Trick 2: import multiple libraries in one line

```python
import os, sys, math
```

## Trick 3: import only what you need

```python
from math import sqrt, pi
```

## Trick 4: use **all** to control wildcard imports

```python
__all__=['func1', 'ClassA']
```

## Trick 5: relative imports in your oun packages

```python
from .module_a import my_function
from ..package_b import utils
```

## Trick 6: use importlib to import dynamically

```python
import importlib
math_module=importlib.import_module("math")
print(math_module.sqrt(4))
```

## Trick 7: Handle import errors gracefully

```python
try:
    import pandas as pd
except ImportError:
    print("Pandas is not installed")
```

## Trick 8: Check what's already imported

```python
import sys
print(sys.modules.keys())
```
