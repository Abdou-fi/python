## Immutable types: - str
- int
- float
- bool
- bytes
- tuple

```python
x = (1, 2) # tuples are immutable
x[0] = 10
```

```python
x = (1, 2) # tuples are immutable
y = x
x = (10, 7) 
print(x)
print(y)
```

```python
x = [1, 2]   # Lists are mutable
y = x
x[0] = 10
print(x)
print(y)
x=[55, 99]
print(y)
# Immutable objects cannot be changed after they are created
# Mutable objects can be changed after they are created
x = "Hello"
print(id(x))
y = x
print(x)
print(y)
x[0] = "h"  # This will raise an error because strings are immutable
x = "hello"
print(id(x))
print(x)
print(y)

# Immutable objects are safer to use in multi-threaded applications because they cannot be changed after they are created

```
