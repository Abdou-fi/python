## Python *"is"* vs. *"=="*
What's the real deal?

### *"is"* operator 
It checks if two variables point to the same object in memory. and compares identity. Often used for ‘None’ (e.g., "x is None”).

### *"=="* operator
It checks if two variables have the same value; and compares values. Most commonly used for comparing numbers, strings, lists, etc.
Relies on the '__eq _' method of objects.
### Example


```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)    # True
print(a is c)    # False
print(a == c)    # True
```
