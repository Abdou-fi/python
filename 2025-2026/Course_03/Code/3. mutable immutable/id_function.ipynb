## <span class="title2">Python Built-in Function: id()</span>

Returns the <span class="orangeText">***unique identity***</span> (memory address) of an object.

### <span class="title3">Syntax: how to use</span>

```python
    id(object)
```
### <span class="title3">Example 1</span>

```python
    x = "hello"
    y = x[0]
    print(id(x))
    print(id(y))
```

````python
x = "hello"
y = x[0]
print(id(x))
print(id(y))
````

<link rel="stylesheet" type="text/css" href="../style.css">

### <span class="title3">Example 2</span>

```python
    a = 42
    b = a
    print(id(a))
    print(id(b))    # same as a, since both reference the same object 
```

````python
a = 42
b = a
print(id(a))
print(id(b))
````

### <span class="title3">When to use it</span>

Use when you want to check if two variables point to the same object.

Helpful for debugging, understanding references, and identity checks.
